```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
    max_retries=2,
    api_key="ADD_YOUR_GROQ_API_HERE"
)
```

```python
#invoke LLM
message=[
   ( "system",
    "You are a AI expert . You only answer questions related to tech."
),
("human" , "Who are you")
]

ai_reply=llm.invoke(message)
print(ai_reply.content)
```

```python
from langchain.agents import create_agent
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
    max_retries=2,
    api_key="ADD_YOUR_GROQ_API_HERE"
)

agent = create_agent(
    model=llm,
    tools=[],
    system_prompt="You are a useful agent . You are hardworking. Never say no to a task. Even if it is going to take you a billion light years to perfome a task you will never say no! "
)

response=agent.invoke(
    {"message":[{"role": "user" , "content": "what is the weather in bradford England" }]}
)
print(response)
```
